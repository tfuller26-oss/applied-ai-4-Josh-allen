### 3. Comparing the formats

**Adaptive Survivor** won the bracket and finished sixth in the round-robin. **My Evolved Agent** won the round-robin. A randomly seeded bracket rewards the ability to beat the specific opponents in a path to the final. In the round-robin, each strategy faces the full field in repeated matches, which gives me a broader measure of how well it performs against this mix of opponents. I would use the bracket to study vulnerability to particular matchups and the round-robin to judge overall scoring performance.

### 4. Awards and trade-offs

King kong earned no superlative in this run. Adaptive Survivor received the Most Aggressive award with a **99.2% win rate**, while My Evolved Agent received Most Cooperative and finished first by total score. I would prioritize total score if the business goal were cumulative profit. Mutual investment pays three points per player per round, so both players can earn substantial returns in a match that ends in a tie. I would still track win rate for situations where relative placement determines who advances or secures a limited opportunity.

### 5. Generalization

Unit 5 trained the strategy against a pool of ten randomly selected opponents. The broader roster here exposed it to a different mix of opening moves, memory windows, and responses to undercutting. Third place suggests the strategy transferred well under this seed, and the bracket loss shows how much a single opponent can matter. If I could redesign the training process, I would hold out some opponent types for validation and evaluate each candidate across multiple seeds. I would also compare total points with performance in its weakest matchups so the genetic algorithm has a reason to retain strategies that recover after a bad start.

The results suggest that strategy quality depends on opponent behavior and on how success is measured. A retailer deciding whether to cut prices faces a similar trade-off.
